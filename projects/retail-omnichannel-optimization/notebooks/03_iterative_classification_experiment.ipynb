```python
# -*- coding: utf-8 -*-
# notebooks/03_iterative_classification_experiment.ipynb
# Eksperymentalna klasyfikacja transakcji przed modyfikacją run_etl.py
# Autor: Paweł Żołądkiewicz, Senior BI Analyst
# Cel: Testowanie reguł klasyfikacji z przesuniętą regułą gratisów

import pandas as pd
import numpy as np
from datetime import datetime
import sys
import os
from pathlib import Path

PROJECT_ROOT = Path.cwd()
# Jeśli uruchamiasz znotebooks/, przejdź poziom wyżej
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Dodatkowe zabezpieczenie: idź w górę, aż znajdziesz run_etl.py
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "run_etl.py").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

sys.path.append(str(PROJECT_ROOT))

# Ścieżki względem root
RAW_PATH = PROJECT_ROOT / "data/raw/online_retail_II.xlsx"
#DOCS_DIR = PROJECT_ROOT / "docs"
#INTERIM_DIR = PROJECT_ROOT / "data/interim"
#DOCS_DIR.mkdir(parents=True, exist_ok=True)
#INTERIM_DIR.mkdir(parents=True, exist_ok=True)


from run_etl import load_raw_excel, clean_raw

print("EKSPERYMENTALNA KLASYFIKACJA TRANSAKCJI")
print("=" * 60)
print("ROI_THRESHOLD = 0.05  # 5% - próg opłacalności")
print("MAX_RULES = 10       # maksymalnie 10 reguł")
print("TARGET_ACCURACY = 0.95  # 95% klasyfikacji")
print("=" * 60)
```

```python
print("KROK 1: Ładowanie i czyszczenie danych...")

raw_data = load_raw_excel(RAW_PATH)
clean_data = clean_raw(raw_data)

print(f"Dane do klasyfikacji: {len(clean_data):,} rekordów")
```

```python
print("KROK 2: Przygotowanie danych do klasyfikacji...")

# Skopiuj dane
df = clean_data.copy()

# Dodaj kolumny klasyfikacyjne
df['TransactionCategory'] = 'ToVerification'
df['ClassificationRule'] = 'None'
df['ClassificationStep'] = 0

print(f"Przygotowano {len(df):,} rekordów do klasyfikacji")
```

```python
def apply_rule(dataframe, rule_name , condition_func, category, step_number):
    """Aplikuje regułę klasyfikacji i zwraca statystyki"""
    mask = (dataframe['TransactionCategory'] == 'ToVerification') & dataframe.apply(condition_func, axis=1)
    count = mask.sum()

    if count > 0:
        dataframe.loc[mask, 'TransactionCategory'] = category
        dataframe.loc[mask, 'ClassificationRule'] = rule_name
        dataframe.loc[mask, 'ClassificationStep'] = step_number

    percentage = (count / len(dataframe)) * 100
    remaining = (dataframe['TransactionCategory'] == 'ToVerification').sum()
    remaining_pct = (remaining / len(dataframe)) * 100

    print(f" {rule_name}: {count:,} rekordów ({percentage:.1f}%)")

    if remaining_pct < 5.0:
        print(f"UWAGA: ToVerification spadło poniżej 5% - rozważ zakończenie iteracji")

    return count, remaining_pct

def show_classification_summary(dataframe):
    """Pokazuje podsumowanie klasyfikacji"""
    summary = dataframe['TransactionCategory'].value_counts().sort_values(ascending=False)
    print("\nPODSUMOWANIE KLASYFIKACJI:")
    for category, count in summary.items():
        percentage = (count / len(dataframe)) * 100
        print(f" {category}: {count:,} ({percentage:.1f}%)")

print("Funkcje pomocnicze zdefiniowane")
```

```python
print("REGUŁA 0: Błędy jakości danych...")

def is_data_error(row):
    
    price_error = row['Price'] == 0.0
    desc_error = (pd.isna(row['Description']) or
                  str(row['Description']).strip() == '' or
                  str(row['Description']).upper() == 'UNKNOWN PRODUCT')

    return price_error or desc_error
                      
count, remaining_pct = apply_rule(df, "Bledy_Jakosci", is_data_error, "DataError", 0)

# --- DOKUMENTACJA REGUŁY 0 ---
print(f"\nREGUŁA 0 - ANALIZA:")
print(f"- Warunek: Price == 0 lub Description brak/nieznane") 
print(f"- Wykryto: {count:,} rekordów ({(count / len(df) * 100):.2f}%)")
print(f"- Pozostało do weryfikacji: {remaining_pct:.2f}%")
print(f"- Kategoria: Błędy jakości danych - wymagają korekty systemowej")
print(f"- Biznes impact: Nie można analizować transakcji bez ceny i opisu")
print("-" * 50)

# Szczegółowa analiza błędów
price_zero_count = (df['Price'] == 0.0).sum()
desc_empty_count = (df['Description'].isna() | 
                   (df['Description'].str.strip() == '') |
                   (df['Description'].str.upper() == 'UNKNOWN PRODUCT')).sum()

print(f"\nRozkład błędów jakości:")
print(f"- Price == 0: {price_zero_count:,} rekordów")
print(f"- Pusty/nieznany opis: {desc_empty_count:,} rekordów")
print(f"- Potencjalne nakładanie: mogą być rekordy z oboma błędami")
```

```python
print("REGUŁA 1: Opłaty pocztowe...")

def is_postage(row):
    stock_code = str(row['StockCode']).upper()
    description = str(row['Description']).upper()
    return (stock_code in ['POST', 'DOT'] or
            'POSTAGE' in stock_code or
            'POSTAGE' in description)

count, remaining_pct = apply_rule(df, "Opłaty_Pocztowe", is_postage, "Postage", 1)

# --- DOKUMENTACJA REGUŁY 1 ---
print(f"\nREGUŁA 1 - ANALIZA:")
#print(f"- Warunek: 
print(f"- Wykryto: {count:,} rekordów ({(count / len(df) * 100):.2f}%)")
print(f"- Pozostało do weryfikacji: {remaining_pct:.2f}%")
print(f"- Kategoria biznesowa: Opłaty pocztowe")
print(f"- Rekomendacja: {'Kontynuuj dodawanie reguł' if remaining_pct > 5 else 'Rozważ zakończenie - cel <5% osiągnięty'}")
```

```python
print("REGUŁA 5: Ręczne korekty księgowe..")

def is_manual_adjustment(row):
    stock_code = str(row['StockCode']).upper()
    description = str(row['Description']).upper()

    manual_cond = (stock_code == 'M' and description == 'MANUAL')
    adjust_cond = (stock_code.startswith('ADJUSTMENT') and description.startswith('ADJUSTMENT BY'))
    adjust_bad_dept = (stock_code == 'B' and description.startswith('ADJUST'))
    
    return manual_cond or adjust_cond

count, remaining_pct = apply_rule(df, "Reczne_Korekty", is_manual_adjustment, "ManualAdjustment", 5)

# --- DOKUMENTACJA REGUŁY 5 ---
print(f"\nREGUŁA 5 - ANALIZA:")
print(f"- Warunek: Manual (M) lub Adjustment* z imionami")
print(f"- Wykryto: {count:,} rekordów ({(count/len(df)*100):.2f}%)")
print(f"- Pozostało: {remaining_pct:.2f}%")
print(f"- Kategoria: Ręczne korekty księgowe")
print(f"- Biznes impact: Wymagają weryfikacji zgodności z polityką")
print(f"- Rekomendacja: {'Kontynuuj dodawanie reguł' if remaining_pct > 5 else 'Rozważ zakończenie - cel <5% osiągnięty'}")
print("-" * 50)
```

```python
print("REGUŁA 6 - Koszty transportu (Carriage)...")

def is_carriage(row):
    stock_code = str(row['StockCode']).upper()
    description = str(row['Description']).upper()
    return (stock_code == 'C2' or description == 'CARRIAGE')

count, remaining_pct = apply_rule(df, "Koszty_Transportu", is_carriage, "TransportCosts", 6)

# --- DOKUMENTACJA REGUŁY 6 ---
print(f"\nREGUŁA X - ANALIZA:")
print(f"- Warunek: StockCode == 'C2' oraz Description == 'CARRIAGE'")
print(f"- Wykryto: {count:,} rekordów ({(count/len(df)*100):.2f}%)")
print(f"- Pozostało: {remaining_pct:.2f}%")
print(f"- Kategoria: Koszty transportu kurierskiego/logistycznego")
print(f"- Biznes impact: Oddzielny od opłat pocztowych - różne dostawcy")
print(f"- Rekomendacja: {'Kontynuuj dodawanie reguł' if remaining_pct > 5 else 'Rozważ zakończenie - cel <5% osiągnięty'}")
print("-" * 50)
    
```

```python
print("REGUŁA 7: Próbki...")

def is_sample(row):
    stock_code = str(row['StockCode']).upper()
    description = str(row['Description']).upper()
    return (stock_code == 'S' or description == 'SAMPLE')

count, remaining_pct = apply_rule(df, "Probki", is_sample, "Sample", 7)

# --- DOKUMENTACJA REGUŁY 7 ---
print(f"\nREGUŁA 7 - ANALIZA:")
#print(f"- Warunek: 
print(f"- Wykryto: {count:,} rekordów ({(count / len(df) * 100):.2f}%)")
print(f"- Pozostało do weryfikacji: {remaining_pct:.2f}%")
print(f"- Kategoria biznesowa: Próbki")
print(f"- Rekomendacja: {'Kontynuuj dodawanie reguł' if remaining_pct > 5 else 'Rozważ zakończenie - cel <5% osiągnięty'}")
```

```python
print("REGUŁA 8: Prawidłowe transakcje...")

def is_valid_sale(row):
    "Regularna sprzedaż - standardowy format kodów i faktur"
    stock_code = str(row['StockCode'])
    invoice = str(row['Invoice'])
    quantity = row['Quantity']

    # StockCode: 5 cyfr lub 5 cyfr + 1 litera
    stock_valid = ((len(stock_code) == 5 and stock_code.isdigit()) or 
                   (len(stock_code) == 6 and stock_code[:5].isdigit() and stock_code[5].isalpha()) or 
                   (len(stock_code) == 7 and stock_code[:5].isdigit() and stock_code[5:].isalpha()))

    # Invoice: tylko cyfry (standardowe numerowanie)
    invoice_valid = invoice.isdigit()

    # Quantity dodatnie
    qty_positive = quantity > 0

    return stock_valid and invoice_valid and qty_positive

count, remaining_pct = apply_rule(df, "Prawidłowe_Transakcje", is_valid_sale, "ValidSales", 8)

# --- DOKUMENTACJA REGUŁY 18 ---
print(f"\nREGUŁA 8 - ANALIZA:")
print(f"- Warunek: StockCode standardowy + Invoice numeryczne + Quantity > 0")
print(f"- Wykryto: {count:,} rekordów ({(count/len(df)*100):.2f}%)")
print(f"- Pozostało: {remaining_pct:.2f}%")
print(f"- Kategoria: Regularna sprzedaż - główne revenue")
print(f"- Biznes impact: Podstawa do analizy KPI sprzedażowych")
print(f"- Rekomendacja: {'Kontynuuj dodawanie reguł' if remaining_pct > 5 else 'Rozważ zakończenie - cel <5% osiągnięty'}")
print("-" * 50)
```

```python
print("REGUŁA 9: Zwroty...")

def is_return(row):
    "Zwroty - faktury Credit Note z ujemnymi ilościami"
    stock_code = str(row['StockCode'])
    invoice = str(row['Invoice'])
    quantity = row['Quantity']

    # StockCode: identycznie jak prawidłowa transakcja
    stock_valid = ((len(stock_code) == 5 and stock_code.isdigit()) or 
                   (len(stock_code) == 6 and stock_code[:5].isdigit() and stock_code[5].isalpha()) or 
                   (len(stock_code) == 7 and stock_code[:5].isdigit() and stock_code[5:].isalpha()))

    # Invoice: zaczyna się od "C" (Credit Note)
    invoice_credit = invoice.upper().startswith('C')
    
    # Quantity ujemne
    qty_negative = quantity < 0

    return stock_valid and invoice_credit and qty_negative

count, remaining_pct = apply_rule(df, "Zwroty", is_return, "Returns", 9)

# --- DOKUMENTACJA REGUŁY 19 ---
print(f"\nREGUŁA 9 - ANALIZA:")
print(f"- Warunek: StockCode standardowy + Invoice 'C*' + Quantity < 0")
print(f"- Wykryto: {count:,} rekordów ({(count/len(df)*100):.2f}%)")
print(f"- Pozostało: {remaining_pct:.2f}%")
print(f"- Kategoria: Zwroty klientów - wpływ na net sales")
print(f"- Biznes impact: Kluczowe dla analizy return rate i customer satisfaction")
print(f"- Rekomendacja: {'Kontynuuj dodawanie reguł' if remaining_pct > 5 else 'Rozważ zakończenie - cel <5% osiągnięty'}")
print("-" * 50)

    
```
